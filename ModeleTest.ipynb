```python
from ultralytics import YOLO
model = YOLO("yolov8s.pt")

results = model.train(
    data="thermal.yaml",
    imgsz=832,
    epochs=100,
    batch=16,
    workers=4,
    device=0,
    patience=20,
)
```

```python
model = YOLO("runs/detect/train/weights/best.pt")
metrics = model.val(data="thermal.yaml", imgsz=640)
print(metrics)
```

```python
model.predict(
    source="data/images/val",
    imgsz=640,
    conf=0.3,
    save=True
)
```
